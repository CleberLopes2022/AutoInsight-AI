```python
data_corte = pd.Timestamp("2025-01-01")

treino = df_ml_v1[
    df_ml_v1["data_referencia"] < data_corte
].copy()

teste = df_ml_v1[
    df_ml_v1["data_referencia"] >= data_corte
].copy()

print(f"Treino: {len(treino):,}")
print(f"Teste: {len(teste):,}")

print(
    f'Período treino: {treino["data_referencia"].min()} '
    f'até {treino["data_referencia"].max()}'
)

print(
    f'Período teste: {teste["data_referencia"].min()} '
    f'até {teste["data_referencia"].max()}'
)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np

# Baseline:
# preço previsto do próximo mês = preço do mês atual
teste["previsao_baseline"] = teste["preco"]

y_real = teste[target]
y_pred = teste["previsao_baseline"]

mae = mean_absolute_error(y_real, y_pred)

rmse = np.sqrt(
    mean_squared_error(y_real, y_pred)
)

r2 = r2_score(y_real, y_pred)

mape = (
    np.abs(
        (y_real - y_pred) / y_real
    ).mean() * 100
)

print("=" * 40)
print("BASELINE — PREÇO ATUAL")
print("=" * 40)

print(f"MAE:  R$ {mae:,.2f}")
print(f"RMSE: R$ {rmse:,.2f}")
print(f"R²:   {r2:.6f}")
print(f"MAPE: {mape:.4f}%")
```

```python
pred_lr = modelo_lr.predict(X_test)
```

```python
mae_lr = mean_absolute_error(
    y_test,
    pred_lr
)

rmse_lr = np.sqrt(
    mean_squared_error(
        y_test,
        pred_lr
    )
)

r2_lr = r2_score(
    y_test,
    pred_lr
)

mape_lr = (
    np.abs(
        (y_test - pred_lr) / y_test
    ).mean() * 100
)

print("=" * 40)
print("LINEAR REGRESSION")
print("=" * 40)

print(f"MAE:  R$ {mae_lr:,.2f}")
print(f"RMSE: R$ {rmse_lr:,.2f}")
print(f"R²:   {r2_lr:.6f}")
print(f"MAPE: {mape_lr:.4f}%")
```

```python
X_train_v4 = treino_v4[features_v4]
y_train_v4 = treino_v4[target_v3]

X_test_v4 = teste_v4[features_v4]
y_test_v4 = teste_v4[target_v3]
```

```python
modelo_hgb_v4 = HistGradientBoostingRegressor(
    learning_rate=0.08,
    max_iter=150,
    max_leaf_nodes=31,
    min_samples_leaf=50,
    l2_regularization=1.0,
    categorical_features="from_dtype",
    random_state=42
)

modelo_hgb_v4.fit(
    X_train_v4,
    y_train_v4
)
```

```python
pred_variacao_v4 = modelo_hgb_v4.predict(
    X_test_v4
)

teste_v4["pred_variacao_hgb_v4"] = (
    pred_variacao_v4
)

teste_v4["preco_prev_hgb_v4"] = (
    teste_v4["preco"]
    * (
        1
        + teste_v4["pred_variacao_hgb_v4"] / 100
    )
)

teste_v4["preco_real_proximo_mes"] = (
    teste_v4["preco"]
    * (
        1
        + teste_v4[target_v3] / 100
    )
)
```

```python
janelas = [
    ("2024", "2024-01-01", "2025-01-01"),
    ("2025", "2025-01-01", "2026-01-01"),
    ("2026", "2026-01-01", "2027-01-01")
]

resultados_temporais = []
```

```python
for nome, inicio, fim in janelas:

    inicio = pd.Timestamp(inicio)
    fim = pd.Timestamp(fim)

    treino = df_ml_v3[
        df_ml_v3["data_referencia"] < inicio
    ].copy()

    teste = df_ml_v3[
        (df_ml_v3["data_referencia"] >= inicio) &
        (df_ml_v3["data_referencia"] < fim)
    ].copy()

    X_train = treino[features_v3]
    y_train = treino[target_v3]

    X_test = teste[features_v3]
    y_test = teste[target_v3]

    modelo = HistGradientBoostingRegressor(
        learning_rate=0.08,
        max_iter=150,
        max_leaf_nodes=31,
        min_samples_leaf=50,
        l2_regularization=1.0,
        random_state=42
    )

    modelo.fit(X_train, y_train)

    pred_var = modelo.predict(X_test)

    preco_real = (
        teste["preco"]
        * (1 + y_test / 100)
    )

    preco_prev = (
        teste["preco"]
        * (1 + pred_var / 100)
    )

    mae = mean_absolute_error(
        preco_real,
        preco_prev
    )

    rmse = np.sqrt(
        mean_squared_error(
            preco_real,
            preco_prev
        )
    )

    mape = (
        np.abs(
            (preco_real - preco_prev)
            / preco_real
        ).mean() * 100
    )

    # baseline: preço atual
    mae_baseline = mean_absolute_error(
        preco_real,
        teste["preco"]
    )

    resultados_temporais.append({
        "Periodo": nome,
        "Treino": len(treino),
        "Teste": len(teste),
        "MAE_Modelo": mae,
        "MAE_Baseline": mae_baseline,
        "RMSE_Modelo": rmse,
        "MAPE_Modelo": mape
    })

    print(f"{nome} concluído")
```

```python
validacao_temporal = pd.DataFrame(
    resultados_temporais
)

validacao_temporal["Ganho_MAE_%"] = (
    (
        validacao_temporal["MAE_Baseline"]
        - validacao_temporal["MAE_Modelo"]
    )
    / validacao_temporal["MAE_Baseline"]
    * 100
)

display(
    validacao_temporal.round(3)
)
```

```python
teste["preco_real"] = preco_real
teste["preco_previsto"] = preco_prev

teste["erro_abs"] = (
    teste["preco_real"]
    - teste["preco_previsto"]
).abs()

teste["erro_pct"] = (
    teste["erro_abs"]
    / teste["preco_real"]
) * 100
```

```python
teste["faixa_preco"] = pd.cut(
    teste["preco"],
    bins=[
        0,
        30000,
        60000,
        100000,
        200000,
        500000,
        float("inf")
    ],
    labels=[
        "Até 30k",
        "30k–60k",
        "60k–100k",
        "100k–200k",
        "200k–500k",
        "500k+"
    ]
)

erro_faixa = (
    teste
    .groupby(
        "faixa_preco",
        observed=True
    )
    .agg(
        Veiculos=("erro_abs", "size"),
        MAE=("erro_abs", "mean"),
        MAPE=("erro_pct", "mean")
    )
    .round(2)
)

display(erro_faixa)
```
